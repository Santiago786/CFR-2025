```python
# create two boolean arrays for if a galaxy is a spiral galaxy or an elliptical galaxy
# use the same probability cut off that we did with the galaxy zoo
# you can assume a galaxy is elliptical if it is is less than 20% chance a spiral galaxy

chance_spiral = data ['Probability is spiral']

galaxy_chance_spiral = chance_spiral >= 0.80
galaxy_elliptical =chance_spiral <= 0.20
print(galaxy_chance_spiral )
print(galaxy_elliptical)
```

```python
print('The number of elliptical galaxies is', sum(galaxy_elliptical)) # fill in the sum after the comma
print('The number of spiral galaxies is', sum(galaxy_chance_spiral)) # fill in the sum after the comma
print('The total number of galaxies classifed is', len(chance_spiral))
```

**What percentage of these galaxies are spirals?**

**Answer:** 26.18%

**Answer:** 11.76%

```python
#fill in the name of the boolean arrays where you see $
print('The percentage of spiral galaxies is', round(sum(galaxy_chance_spiral)/len(chance_spiral) * 100,2), '%')
print('The percentage of elliptical galaxies is', round(sum(galaxy_elliptical)/len(chance_spiral) * 100,2), '%')
```

**How do these percentages compare to the ones you found in the Galaxy Zoo data?**

**Answer:** Very similiar

```python
# create a single plot showing histograms of the colors of the two types of galaxies 

# import our graphing package and numpy
import numpy as np

#the below is optional but makes the font a size I can read more easily
import matplotlib.pyplot as plt
matplotlib.rc('font', **{'family':'sans-serif', 'size':16}) 

# calculate color
g_band = data['g band']
r_band = data['r band']

color = g_band - r_band



# separate into spiral and elliptical
plt.hist(color[galaxy_chance_spiral], bins=75, color= 'teal', edgecolor = "white")
plt.xlabel('color')
plt.ylabel('Illustriss Galaxy')

plt.hist(color[galaxy_elliptical], bins=75, color= 'black', edgecolor = "r")
plt.xlabel('color')


# create one histogram with elliptical and spiral galaxies on it
mean_spirals = np.mean(galaxy_chance_spiral)
mean_ellipticals = np.mean(galaxy_elliptical)
print("Arithmetic Mean Spiral:", mean_spirals)
print("Arithmetic Mean Ellip:", mean_ellipticals)

# statistics -- find the mean and standard deviations of the color for spirals and ellipticals
# print the stats below
```
